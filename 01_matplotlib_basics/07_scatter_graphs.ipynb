Scatter Graph

Shows the relationship between two variables


```python
import matplotlib.pyplot as plt
import numpy as np
```

<!-- Shows Relationship between two varibles
Helps to identify a correlation (+,-,None)
Example "Study Hours vs Test Scores" -->

```python
x1 = np.array([0,1,1,2,3,4,5,6,7,7,8])
y1= np.array([55,60,65,62,68,70,75,78,82,85,87])

x2 = np.array([0,1,2,2,3,4,5,6,7,8,8])
y2= np.array([45,50,65,26,8,70,78,83,83,88,92])


plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Test Scores")
plt.scatter(x1,y1,color = "Purple",
alpha =0.5,s = 100,label = "Class A")#s is size
#alpha is transparency


plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Test Scores")
plt.scatter(x2,y2,color = "red",
alpha =0.5,s = 100,label = "Class B")#s is size
#alpha is transparency

plt.legend()#Needed to show the legend
```

Points moving up and to the right = Positive correlation (one goes up, the other goes up).

Points moving down and to the right = Negative correlation (one goes up, the other goes down).

Points scattered randomly with no shape = No correlation.
