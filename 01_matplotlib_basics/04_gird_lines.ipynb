```python
#grid() - Helps make plot easier to read by adding reference lines.

x = [1,2,3,4,5]
y = [5,10,15,20,25]

plt.grid(axis= "x",linewidth =2,color ='pink',linestyle = 'dashdot') 
#adds the grid lines
#x-axis - only x axis
#y-axis - only y axis
plt.plot(x,y)
```

Q2. Full chart
Using the same data, make a chart with:

Circle markers, markersize 10, dashed blue line
A title "Sales Trend" (bold, fontsize 20) and axis labels "Day" and "Sales"
xticks showing only the values in x
A grid on both axes: dashdot, gray, linewidth 1

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 9, 6, 15, 12])

plt.plot(x,y,marker = 'o',
         linestyle = "dashed",
         color = "blue",
         ms = 10)

plt.title("Sales Trend",fontweight = "bold",
          fontsize = 20)

plt.xlabel("Day")
plt.ylabel("Sales")

plt.xticks(x)

plt.grid(axis="both", color="green", linestyle="dotted", linewidth=1)


```
