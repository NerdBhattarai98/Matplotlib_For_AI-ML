Bar Charts 

```python
import matplotlib.pyplot as plt
import numpy as np
```

#Bar chart = compare categories of data by representing each category with a bar diagram

```python
categories = np.array(["Grains","Fruits","Vegetables","Protein","Diary","Sweets"])
values =np.array([4,3,2,5,3,1])

#plt.bar(categories,values,color = "purple") #first pass x axis then y

plt.barh(categories,values,color = "purple") #Horizontal BarCharts
plt.title("Daily Consumptions")
plt.grid()

plt.xlabel("Food")
plt.ylabel("Quantity")

# Axis labels are swapped for barh
```

```python
# Part 1: vertical
plt.bar(categories, values, color="teal")
plt.title("Weekly Eating Hours", fontsize=18, fontweight="bold")
plt.xlabel("Food")      # categories on x
plt.ylabel("Quantity")        # values on y
plt.grid(axis="y", linestyle="dashed", color="gray")
plt.show()

# Part 2: horizontal
plt.barh(categories, values, color="orange")
plt.title("Weekly Eating Hours", fontsize=18, fontweight="bold")
plt.xlabel("Food")        # values now on x
plt.ylabel("Quantity")      # categories now on y
plt.grid(axis="x", linestyle="dashed", color="gray")
plt.show()
```
