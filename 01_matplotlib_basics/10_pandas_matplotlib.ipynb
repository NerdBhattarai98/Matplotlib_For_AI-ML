```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv(r"C:\Users\Anbit Bhattarai\Desktop\Matplotlib_For_AI-ML\01_matplotlib_basics\pokemon_with_bmi.csv")
print(df)

```

```python
print(df["Type1"])
print(df["Type1"].value_counts())
```

```python
type_count = df["Type1"].value_counts(ascending = True)

plt.barh(type_count.index,type_count.values,color = "lightgreen",ec = "black")
plt.tight_layout()

plt.title("Number of Pokemon by Primary Type")
plt.xlabel("Counts")
plt.ylabel("Pokemon")
plt.grid()
```
