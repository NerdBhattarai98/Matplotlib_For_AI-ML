Pie Chart

Pie Chart = Circular Chart Divided into slices to show percentage of total.
Good for visualizing distribution among categroies


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories = np.array(["Freshman","Sophomores","Juniors","Seniors"])
values = np.array([300,250,275,225])
colors = ["red","blue","green","yellow"]
plt.pie(values,labels=categories,autopct="%1.1f%%",colors=colors,explode=[0.6,0,0,0.1],shadow=True,startangle=180 ) #Double %% because % is only for formate specifier
#Explodes the senior 0.1
plt.title("My College")
```
