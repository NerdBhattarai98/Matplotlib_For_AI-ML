```python
import numpy as np
import matplotlib.pyplot as plt
```

Histograms

Histogram = A visual representation of the distribution of quantitative data.
thet group values into bais( intervals)
and counts how many falls in each range

```python
scores = np.random.normal(loc = 80 , scale = 10, size=1000) 
#loc -data will be around the 90 marks
#scales - How scattred do you want your data to be
#size = nu,mber of elements
scores = np.clip(scores,50,100)
# plt.grid()

plt.hist(scores,bins = 10,color = "lightgreen",ec = "black")
plt.title("Exam Scores")
plt.xlabel("Score")
```

Lower scale (e.g., scale = 2): The data points will cluster very tightly around the mean (80), resulting in a tall, narrow bell curve on your histogram.

Higher scale (e.g., scale = 20): The data points will be much more spread out across a wider range of values, resulting in a short, wide, and flattened bell curve.

scale=10 means most of the generated values will typically fall within +- 10 units of the mean (roughly between 70 and 90, before clipping).
